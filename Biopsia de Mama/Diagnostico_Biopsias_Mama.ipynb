```python
import numpy as np
import pandas as pd
import joblib
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

RANDOM_STATE = 42
rng = np.random.RandomState(RANDOM_STATE)
```

### Constantes y Funciones Movedoras (para asegurar el orden correcto)

```python
COLUMNAS_PREDICTORAS = [
    'mean texture', 'mean smoothness', 'mean symmetry', 'mean fractal dimension',
    'texture error', 'smoothness error', 'symmetry error',
    'variabilidad_por_biopsia', 'num_biopsias_previas'
    # 'sesiones_tratamiento_programadas', # Esta columna introduce data leakage
]
```

```python
def construir_caracteristicas(df):
    """Agrega una métrica derivada de variabilidad de textura por biopsia previa.

    OJO: esta columna puede dar inf cuando num_biopsias_previas == 0 -- el arreglo
    (reemplazar por NaN e imputar) se hace en entrenar_modelo, DESPUÉS del split y
    usando solo la mediana de train, para no filtrar información del test hacia el
    entrenamiento (ver sección 5)."""
    df = df.copy()
    df['variabilidad_por_biopsia'] = df['mean texture'] / df['num_biopsias_previas']
    return df
```

## 1. Carga de datos

Cargamos el dataset base y lo enriquecemos con dos columnas que el equipo clínico pidió agregar:

- `num_biopsias_previas`: cuántas biopsias se le habían hecho antes a la paciente.
- `sesiones_tratamiento_programadas`: cuántas sesiones de tratamiento tiene ya programadas la paciente.

```python
def cargar_datos():
    """Carga el dataset de biopsias y agrega las columnas clínicas complementarias."""
    data = load_breast_cancer(as_frame=True)
    df = data.frame.copy()
    df['diagnostico'] = data.target  # 0 = maligno, 1 = benigno
    n = len(df)

    df['num_biopsias_previas'] = rng.randint(0, 4, size=n)
    df['sesiones_tratamiento_programadas'] = np.where(
        df['diagnostico'] == 0,
        rng.randint(3, 9, size=n),
        0,
    )
    return df


df = cargar_datos()
df[['mean texture', 'mean smoothness', 'num_biopsias_previas',
    'sesiones_tratamiento_programadas', 'diagnostico']]
```

## 3. Validación de calidad de datos

Un chequeo simple antes de entrenar: que no haya valores infinitos ni nulos colándose en las columnas numéricas.

```python
def validar_datos(df, columnas):
    """Revisa que las columnas indicadas no tengan nulos ni valores infinitos."""
    problemas = {}
    for col in columnas:
        n_nulos = df[col].isna().sum()
        n_infinitos = np.isinf(df[col]).sum() if np.issubdtype(df[col].dtype, np.number) else 0
        if n_nulos or n_infinitos:
            problemas[col] = {'nulos': int(n_nulos), 'infinitos': int(n_infinitos)}
    return problemas

# --- Diagnóstico del ValueError: ¿dónde nace el valor no finito? ---
df_crudo = cargar_datos()
print("1) Recién cargado: ",
      validar_datos(df_crudo, ['mean texture', 'num_biopsias_previas']))

df_derivado = construir_caracteristicas(df_crudo)
print("2) Con la derivada:",
      validar_datos(df_derivado, COLUMNAS_PREDICTORAS))

malas = df_derivado[~np.isfinite(df_derivado['variabilidad_por_biopsia'])]
print("3) Filas no finitas:", len(malas))
print(malas['num_biopsias_previas'].value_counts())

```

## 4. Ingeniería de características

A partir de la textura promedio y del número de biopsias previas, construimos un indicador de variabilidad por biopsia — mientras más biopsias previas tenga la paciente, más se "diluye" la variabilidad reportada.

## 5. Entrenamiento del modelo

Usamos regresión logística — el mismo clasificador que ya conoces — sobre un subconjunto de columnas clínicas y de imagen.

```python
def entrenar_modelo(df, usar_class_weight=False):
    """Entrena un clasificador de regresión logística y devuelve el modelo, sus métricas
    y las predicciones de test."""
    X = df[COLUMNAS_PREDICTORAS]
    y = df['diagnostico']
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.25, random_state=RANDOM_STATE, stratify=y
    )

    # Corrección: variabilidad_por_biopsia puede ser inf (num_biopsias_previas == 0).
    # Se reemplaza por NaN y se imputa con la mediana calculada SOLO en train, para no
    # filtrar información del test hacia el entrenamiento.
    X_train = X_train.copy()
    X_test = X_test.copy()
    X_train['variabilidad_por_biopsia'] = X_train['variabilidad_por_biopsia'].replace([np.inf, -np.inf], np.nan)
    X_test['variabilidad_por_biopsia'] = X_test['variabilidad_por_biopsia'].replace([np.inf, -np.inf], np.nan)
    mediana_train = X_train['variabilidad_por_biopsia'].median()
    X_train['variabilidad_por_biopsia'] = X_train['variabilidad_por_biopsia'].fillna(mediana_train)
    X_test['variabilidad_por_biopsia'] = X_test['variabilidad_por_biopsia'].fillna(mediana_train)

    kwargs = {'max_iter': 5000}
    if usar_class_weight:
       kwargs['class_weight'] = 'balanced'

    modelo = LogisticRegression(**kwargs)
    modelo.fit(X_train, y_train)

    acc_train = accuracy_score(y_train, modelo.predict(X_train))
    y_pred = modelo.predict(X_test)
    acc_test = accuracy_score(y_test, y_pred)
    return modelo, acc_train, acc_test, y_test, y_pred
```

## 6. Ejecutar el pipeline completo

Esto es lo que corre en producción: carga los datos, construye las características y entrena el modelo.

```python
df = cargar_datos()
df = construir_caracteristicas(df)
modelo, acc_train, acc_test, y_test, y_pred = entrenar_modelo(df, usar_class_weight=True)
print(f"Accuracy en entrenamiento: {acc_train:.4f}")
print(f"Accuracy en prueba:        {acc_test:.4f}")
```

```python
from sklearn.metrics import f1_score, precision_score, recall_score, confusion_matrix, ConfusionMatrixDisplay

# Métrica global (clase 1 = benigno, la que usa sklearn por defecto)
f1_benigno = f1_score(y_test, y_pred)
print(f"F1-score (benigno, clase por defecto): {f1_benigno:.4f}")
print()

# Métricas de la clase "maligno" (0) -- la que importa clínicamente detectar
precision_maligno = precision_score(y_test, y_pred, pos_label=0)
recall_maligno = recall_score(y_test, y_pred, pos_label=0)
f1_maligno = f1_score(y_test, y_pred, pos_label=0)

print(f"Precision (maligno):     {precision_maligno:.4f}")
print(f"Exhaustividad (maligno): {recall_maligno:.4f}")
print(f"F1 (maligno):            {f1_maligno:.4f}")
```

```python
cm = confusion_matrix(y_test, y_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['maligno (0)', 'benigno (1)'])
disp.plot(cmap='Blues', values_format='d')
plt.title('Matriz de Confusión')
plt.show()
```

### Interpretación de la Matriz de Confusión (corregida)

`confusion_matrix(y_test, y_pred)` ordena las clases como `[0, 1]` = `[maligno, benigno]`. Las filas son la clase **real**, las columnas la clase **predicha**:

|                        | Predicho: maligno (0) | Predicho: benigno (1) |
|---|---|---|
| **Real: maligno (0)**  | Verdadero maligno (TP de detección de cáncer) | **Falso negativo** — cáncer real que el modelo dice "benigno" |
| **Real: benigno (1)**  | **Falso positivo** — benigno que el modelo marca como "maligno" | Verdadero benigno |

Tratando "maligno" como la clase de interés clínico (lo que se quiere detectar), no como la clase "positiva" por defecto de sklearn (que es 1 = benigno):

- La celda **superior-izquierda** son los tumores malignos que el modelo detectó correctamente.
- La celda **superior-derecha** son **falsos negativos de malignidad**: pacientes con cáncer real a quienes el modelo les dice "benigno". Es el error más grave posible en este contexto — la paciente no sería referida a tiempo.
- La celda **inferior-izquierda** son **falsos positivos de malignidad**: pacientes sin cáncer a quienes el modelo marca como "maligno". Genera pruebas adicionales y ansiedad innecesaria, pero no pone en riesgo la vida de la paciente.
- La celda **inferior-derecha** son los tumores benignos correctamente identificados.

(La versión anterior de esta celda tenía las etiquetas TN/FP/FN/TP invertidas respecto al error clínicamente relevante, y se contradecía a sí misma entre el cuerpo del texto y el resumen final — se corrigió aquí.)

### ¿Qué métrica es prioritaria?

Los dos tipos de error de este modelo no cuestan lo mismo. Un **falso negativo de malignidad** (el modelo dice "benigno" cuando el tumor en realidad es maligno) significa que una paciente con cáncer real no es referida a tiempo para tratamiento — el costo es potencialmente su vida, y crece cuanto más tiempo pase sin detectarse. Un **falso positivo de malignidad** (el modelo dice "maligno" cuando en realidad es benigno) tiene un costo real — pruebas adicionales, ansiedad, tiempo del especialista — pero es un costo recuperable, no potencialmente mortal. Como esta aplicación solo ayuda a *priorizar* qué casos revisa primero un especialista (la decisión final siempre es humana), el peor escenario no es generar una alarma de más, sino dejar pasar sin marcar un caso maligno real. Por eso la **exhaustividad (recall) de la clase "maligno"** es la métrica prioritaria en este contexto: mide directamente qué proporción de los tumores malignos reales el modelo efectivamente detecta, y es exactamente el complemento del error más costoso (1 − recall = tasa de falsos negativos de malignidad). La precisión y el F1 de "maligno" siguen siendo útiles como contexto — un recall perfecto marcando todo como maligno no serviría de nada en la práctica, generaría demasiadas falsas alarmas — pero ante un trade-off entre subir accuracy/precisión y mantener el recall de "maligno", se prioriza no sacrificar el recall.

### Peso de cada variable en el modelo

En regresión logística, cada coeficiente indica cuánto empuja esa variable hacia "benigno" o hacia "maligno". Una variable con un peso muchísimo mayor que las demás merece una segunda mirada.

```python
coeficientes = pd.Series(modelo.coef_[0], index=COLUMNAS_PREDICTORAS).sort_values()

plt.figure(figsize=(7, 5))
colores = ['#c0392b' if v < 0 else '#2980b9' for v in coeficientes]
coeficientes.plot(kind='barh', color=colores)
plt.axvline(0, color='black', linewidth=0.8)
plt.title('Coeficientes de la regresión logística')
plt.xlabel('Peso (negativo empuja hacia maligno, positivo hacia benigno)')
plt.tight_layout()
plt.show()
```

## 8. Predecir un caso nuevo

Así es como el equipo clínico piensa usar la app: le pasan las mediciones de una paciente nueva y el modelo devuelve una predicción.

```python
def predecir_caso(modelo, caso: dict):
    """Predice el diagnóstico para una paciente nueva a partir de un diccionario de mediciones."""
    entrada = pd.DataFrame([caso])[COLUMNAS_PREDICTORAS]
    pred = modelo.predict(entrada)[0]
    return 'benigno' if pred == 1 else 'maligno'


caso_nuevo = {
    'mean texture': 18.2,
    'mean smoothness': 0.095,
    'mean symmetry': 0.18,
    'mean fractal dimension': 0.062,
    'texture error': 1.1,
    'smoothness error': 0.006,
    'symmetry error': 0.02,
    'variabilidad_por_biopsia': 18.2,
    'num_biopsias_previas': 1,
}

predecir_caso(modelo, caso_nuevo)
```
